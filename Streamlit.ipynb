```python
import ipywidgets as widgets
from IPython.display import display, HTML

# Membuat form interaktif di dalam Jupyter
ip_input = widgets.IntText(value=1000, description='IP ID:')
app_input = widgets.IntText(value=10, description='App ID:')
threshold_slider = widgets.FloatSlider(value=0.7, min=0.5, max=0.95, step=0.05, description='Threshold:')
output_box = widgets.Output()

def on_button_clicked(b):
    with output_box:
        output_box.clear_output()
        print(f"Memproses data dengan IP: {ip_input.value}, App: {app_input.value}...")
        print(f"Confidence Threshold: {threshold_slider.value}")
        print("Status: [AMAN] Transaksi valid (Bukan Fraud).")

button = widgets.Button(description="Analisis Transaksi", button_style='success')
button.on_click(on_button_clicked)

# Tampilkan di notebook
display(HTML("<h3>🛡️ Dashboard Interaktif Deteksi Fraud</h3>"))
display(ip_input, app_input, threshold_slider, button, output_box)
```
